### 43. Ek simple calculator banayein (+, -, *, /) symbols ke basis par operation perform karein.

```python
# Write your code here
num1 = float(input("Enter value1"))
op = input("Enter operator")
num2 = float(input("Enter value2"))
if op == "*":
    result = num1 * num2
elif op == "/":
    result = num1 / num2
elif op == "+":
    result = num1 + num2
elif op == "-":
    result = num1 - num2
else:
    result = "invalid operation"
print(result)
```
